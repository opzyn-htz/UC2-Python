```python
#Desafio Integrador Equipe Glaucio e Matheus

TIPOS = ("notebook", "tablet", "kit arduino", "fone") 
equipamentos = [ 
{"codigo": "NB01", "tipo": "notebook", "disponivel": True, 
"com": "", "emprestimos": 4}, 
{"codigo": "NB02", "tipo": "notebook", "disponivel": False, 
"com": "Ana", "emprestimos": 7}, 
{"codigo": "TB01", "tipo": "tablet", "disponivel": True, 
"com": "", "emprestimos": 2}, 
{"codigo": "AR01", "tipo": "kit arduino", "disponivel": False, 
"com": "Bruno", "emprestimos": 5}, 
{"codigo": "FN01", "tipo": "fone", "disponivel": True, 
"com": "", "emprestimos": 0}, ]
```

```python
def listar():
    print(*equipamentos,sep='\n')
```

```python

def cadastrar():
    equipamento=input("Tipo de equipamento: ")
    codigo=input("Código novo: ").upper()
    if equipamento not in TIPOS:
        print("Tipo de equipamento inválido!")
        return cadastrar()

    for v in equipamentos:
        if v["codigo"] != codigo:
            pass
        else:
            print("Código já cadastrado!")
            return cadastrar()
    disponivel=True
    com=""
    emprestimos=0
    novoitem={"codigo":codigo,"tipo":equipamento,"disponivel":disponivel,"com":com,"emprestimos":emprestimos}
    equipamentos.append(novoitem)


    print(novoitem)
```

```python
def verificar(): # Verifica se o usuário tem mais de 2 equipamentos emprestados, caso tenha, não permite emprestar mais.
    nome=input("Digite seu nome: ").lower()
    emprestimos=0
    for v in equipamentos:
        if v["com"].lower()==nome:
            emprestimos+=1
    if emprestimos>=2:
        print("Você já possui 2 equipamentos emprestados, não é possível realizar mais empréstimos.")
        return False
    else:
        print("Você possui ",emprestimos," equipamentos emprestados, é possível realizar mais empréstimos.")
        return True
```

```python
def emprestar():
    if verificar()==False:
        return "Não pode pegar mais equipamentos", False
    else:
        pass
    codigo=input("Código do equipamento: ").upper()

    posdic=0
    
    while equipamentos[posdic]["codigo"]!=codigo:
         posdic+=1
    
    if equipamentos[posdic]["disponivel"]!=False:
        pass
    else:
        print("Equipamento em Empréstimo!")
    if codigo in equipamentos[posdic]["codigo"]:
        if equipamentos[posdic]["disponivel"]==True:
            equipamentos[posdic]["disponivel"]=False
            equipamentos[posdic]["com"]=input("Nome do emprestimo")
            equipamentos[posdic]["emprestimos"]+=1
        else:
            print("Já emprestado!")
    else:
        print("Equipamento não encontrado")
        
```

```python
def verificar():
    nome=input("Digite seu nome: ").lower()
    emprestimos=0
    for v in equipamentos:
        if v["com"].lower()==nome:
            emprestimos+=1
    if emprestimos>=2:
        print("Você já possui 2 equipamentos emprestados, não é possível realizar mais empréstimos.")
        return False
    else:
        print("Você possui ",emprestimos," equipamentos emprestados, é possível realizar mais empréstimos.")
        return True
```
